# Baseline

```python
import Modules.Data as data
import numpy as np

baseline_filepath = "Data/Output-Baseline.csv"
heating_column_title = "Energy_Heating(kW)"
gas_price = 0.0273 # per kWh

gas_usage = data.column_from_csv(baseline_filepath, heating_column_title)

annual_gas_cost = np.sum(gas_usage) * gas_price

print("Heating gas cost $" + str(round(annual_gas_cost,2)))

cooling_column_title = "Energy_Cooling(kW)"

electricity_price = []

for i in range(6):
    electricity_price.append(0.1)
    
for i in range(6, 23):
    electricity_price.append(0.145)

for i in range(23, 24):
    electricity_price.append(0.1)
    
electricity_price = electricity_price * 365

electricity_usage = data.column_from_csv(baseline_filepath, cooling_column_title)

annual_electricity_cost = np.sum(electricity_usage * electricity_price)

print("Cooling electricity cost $" + str(annual_electricity_cost))
```
