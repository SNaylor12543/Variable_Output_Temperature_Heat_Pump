# Baseline

```python
import Modules.Data as data
import matplotlib.pyplot as plt
import numpy as np

baseline_filepath = "Data/NPV/Baseline.csv"
year_column_title = "Year"
NPV_column_title = "Net Present Value"

Year = data.column_from_csv(baseline_filepath, year_column_title)
NPV = data.column_from_csv(baseline_filepath, NPV_column_title) / 1e6

plt.plot([30, 30], [0, 2.1], '--r', label="30 HVAC year life time")

plt.plot([50, 50], [0, 2.1], '--b', label="50 Building year life time")

plt.plot(Year, NPV, label="baseline")
plt.xlabel('Year')
plt.ylabel('NPV (Millions of dollars)')
plt.legend()
plt.title("How Net Present Value Changes With System Lifetime")
```
